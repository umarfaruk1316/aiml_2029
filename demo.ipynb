# simple calculator program
3 input from user
int, str, int

```python
a = int(input())
op = input() # string data type +,-,*  (type)
b = int(input())

if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print( a * b)
elif op == "/":
    print( a / b)
elif op == "**":
    print( a ** b)
elif op == "%":
    print( a % b)
elif op =="//":
    print(a // b)
else:
    print("invalid operator")
```
