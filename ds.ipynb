# Separte the Postive and negative into a list
# +ve = []
# -ve = []

```python
numbers = [1,24,-2,34,65,0,-34,-3,45,5,8,10,-15]
pos = []
neg = []

for i in numbers:
    if i>=0:
        pos.append(i)
    else:
        neg.append(i)
print("+ve list :",pos)
print("-ve list :", neg)

a = sorted(pos)
print("second largets :", a[-2])
b = sorted(neg)
print("last element :", b[-1])
```

# s = {1,3,5,7,9,11,13}

# input:
6
pop - ['pop']
print - ['print']
add 10  - ['add',10]
discard 5 - ['discard',5]
discard 9 - ['discard',5]
remove 11 - ['remove',11]

```python
n = int(input()) # 6
s = {1,3,5,7,9,11,13}
for i in range(n):
    m = input().split() #pop,print,add 10
    if m[0]=='pop':
        s.pop()
    elif m[0]=='print':
        print(s)
    elif m[0]=='remove':
        s.remove(m[1])
    elif m[0]=='discard':
        s.discard(m[1])
```
