password = input() # Aiml@1234

1. atleast 8 chars
2. atleast 1 uppercase
3. atleast 1 lowercase
4. atleast 1 spl char
5. atleast 1 digit

```python
password = "AimlCse113"

upper_let = False
lower_let = False
digit = False
spl = False
spl_chr = "!@#$%^&*()"

for i in password:
    if len(password) < 8: # base conditation
        print("invalid password")
    else:
        if i.isupper():
            upper_let = True
        elif i.islower():
            lower_let = True
        elif i.isdigit():
            digit = True
        elif i in spl_chr:
            spl = True

if upper_let and lower_let and digit and spl:
    print("valid password")
else:
    print("invalid password")
```
