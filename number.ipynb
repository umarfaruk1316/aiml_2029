# prime or Not

```python
n = int(input())

is_prime = True

if n<1:
    print("Not prime")
else:
    for i in range(2,int(n**0.5)+1): # range(2,int(n**0.5))
        if n%i==0:  # 29%2, 29%3 ,29%4
            is_prime = False
        
if is_prime: # True or False
    print("Prime Number")
else:
    print("Not prime Num")
```

# armstrong number

```python
n = int(input()) # 153
no_of_digits = len(str(n)) # lenth
total = 0

for i in str(n): # 1 * 3
    power = int(i)** int(no_of_digits)
    total += power

if total==n:
    print("Armstrong Num")
else:
    print("not")
```
