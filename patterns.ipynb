# hallow Square

```python
n = int(input()) 

for i in range(n):
    if i==0 or i ==n-1:
        print("* " * n)
    else:
        print("* " + " " * (n+1) + " *")
```

```python
# hallow triangle
```

```python
n = int(input())

for i in range(n):
    if i == 0:
        print(" "*(n-1) + "*")
    elif i == n-1:
        print("* " * n)
    else:
        ls = " " * (n-i-1)
        hs = " " * 2*(i-1) 
        print(ls + "*" + hs + " *")
        
```

```python
n = int(input())

for i in range(1,n+1):
    if i == 1:
        print("* " * n)
    elif i == n:
        print(" " * n + "*")
    else:
        ls = " "
        hs = "  " * (n-i-1)
        print(ls + "*" + hs + "*")
```

```python
n = int(input())

for i in range(1,n+1): # i = 1,2
    print(" " * (n-i), end=" ")
    for j in range(1,i+1):
        print(j, end=" ")

    print()
```
