## 1. Variables, Types, and Input

- Valid names start with a letter or `_`; they cannot contain spaces or start with a digit.
- Main types: `int`, `float`, `str`, `bool`. Check with `type(value)`.
- `input()` always returns a string. Convert before arithmetic.

- `None` means no value; do not confuse it with `0`, `False`, or `""`.

## 2. Operators and Conditions

| Operator | Meaning |
|---|---|
| `+ - * /` | arithmetic; `/` gives float |
| `//` | floor division |
| `%` | remainder; useful for evenness and cycles |
| `**` | exponent |
| `== != < <= > >=` | value comparison |
| `and or not` | boolean logic |

- Use `==` to compare values. Use `is` mainly for identity checks such as `value is None`.
- Conditions can be simplified: `if 0 <= marks <= 100:`.
- Put mutually exclusive cases in `if / elif / else`; order ranges from most specific to most general.
- Short-circuiting: `A and B` skips `B` when `A` is false; `A or B` skips `B` when `A` is true.

```python
number = int(input())
if number > 0:
    result = "positive"
elif number < 0:
    result = "negative"
else:
    result = "zero"
print(result)

# Leap year: divisible by 400, or divisible by 4 but not by 100
def is_leap_year(year):
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)
```

## 3. Strings

- Strings are ordered, indexable, and **immutable**. Slicing creates a new string.
- Indexing: `text[0]`, `text[-1]`; slicing: `text[start:stop:step]`. Stop is exclusive.
- Useful methods: `lower`, `upper`, `strip`, `split`, `join`, `replace`, `count`, `find`, `isdigit`, `isalpha`, `isalnum`.
- Membership is direct and readable: `character in "aeiou"`.
- Interview trick: normalize first with `text = text.strip().lower()`, then solve.
- Use `''.join(parts)` instead of repeated string concatenation in a large loop.

```python
text = "  Python Programming  "
clean = text.strip().lower()
words = clean.split()
reversed_text = clean[::-1]
vowels = sum(character in "aeiou" for character in clean)
print(words, reversed_text, vowels)

# Palindrome check after normalization
def is_palindrome(value):
    value = value.lower().replace(" ", "")
    return value == value[::-1]
```

## 4. Loops and the Problem-Solving Template

- `for` is best when iterating over a sequence or known range.
- `while` is best when repetition depends on a changing condition. Always update the state.
- `range(start, stop, step)` excludes `stop`; `range(n)` means `0` through `n - 1`.
- Use `break` to stop early and `continue` to skip an iteration.
- Accumulator pattern: initialize before the loop, update inside, return or print after.
- Counter pattern: count only when a condition is true.
- Nested loops often mean $O(n^2)$; ask whether sorting, a set, or a dictionary can remove one loop.

```python
numbers = [3, 8, 2, 9, 4]
total = 0
even_count = 0
for number in numbers:
    total += number
    if number % 2 == 0:
        even_count += 1
print(total, even_count)

index = 0
while index < len(numbers):
    print(numbers[index])
    index += 1
```

## 5. Number Problems and High-Value Tricks

- Even/odd: `n % 2 == 0`.
- Last digit: `n % 10`; remove last digit: `n // 10`.
- Prime check: test divisors only from `2` through `isqrt(n)`. A factor larger than the square root has a matching smaller factor.
- Armstrong number: sum each digit raised to the number of digits.
- For digit problems, decide whether to use arithmetic (`%`, `//`) or string conversion. Strings are often simpler; arithmetic avoids conversion.
- Do not forget edge cases: `n < 2` is not prime, zero has one digit, and negative input may need a stated rule.
- For money breakdown, repeatedly use quotient for the count and remainder for the next amount.

```python
from math import isqrt

def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, isqrt(number) + 1):
        if number % divisor == 0:
            return False
    return True

def is_armstrong(number):
    digits = str(number)
    power = len(digits)
    return sum(int(digit) ** power for digit in digits) == number

print(is_prime(29), is_armstrong(153))
```

## 6. Pattern Problems

- The outer loop usually controls rows; the inner loop controls items in a row.
- Ask three questions: how many rows, how many symbols per row, how many leading spaces?
- Common row formulas: solid square has `n` symbols each row; a right triangle has `row` symbols; an inverted triangle has `n - row + 1`.
- Prefer string repetition for simple patterns: `"* " * count`.
- Use `end=" "` when printing on one line, then a blank `print()` after each row.
- Interview tip: write the expected output for `n = 3` before coding. It exposes off-by-one errors quickly.

```python
def right_triangle(size):
    for row in range(1, size + 1):
        print("* " * row)

def number_triangle(size):
    for row in range(1, size + 1):
        print(" " * (size - row), end="")
        for number in range(1, row + 1):
            print(number, end=" " )
        print()

right_triangle(4)
number_triangle(4)
```

## 7. Lists

- Lists are ordered, mutable, and allow duplicates. Indexing is $O(1)$; searching is usually $O(n)$.
- Core operations: `append`, `extend`, `insert`, `remove`, `pop`, `sort`, `reverse`, `count`, `index`.
- `sort()` changes the list; `sorted(list)` returns a new list.
- Slicing: `items[start:stop:step]`; reverse with `items[::-1]`.
- `append(x)` adds one object; `extend(values)` adds each item.
- Avoid aliasing: `copy = original[:]` or `original.copy()`; `copy = original` points to the same list.
- List comprehensions are ideal for simple transform/filter tasks; use a normal loop when logic becomes hard to read.

```python
numbers = [1, -2, 4, -5, 8, 0]
positives = [number for number in numbers if number > 0]
squares = [number * number for number in numbers]

# Rotate left by k positions; modulo handles k > len(items)
def rotate_left(items, k):
    if not items:
        return []
    k %= len(items)
    return items[k:] + items[:k]

print(positives, squares, rotate_left([0, 1, 2, 3, 4], 2))
```

## 8. Tuples, Sets, and Dictionaries

### Tuple
- Ordered and immutable; useful for fixed records and safe unpacking.
- A one-item tuple needs a comma: `(value,)`.

### Set
- Unordered collection of unique hashable values. Average membership is $O(1)$.
- Use for deduplication, fast membership, and set operations: union `|`, intersection `&`, difference `-`.

### Dictionary
- Key-value mapping; keys must be unique and hashable. Average lookup, insert, and delete are $O(1)$.
- Counting pattern: `counts[item] = counts.get(item, 0) + 1`.
- Grouping pattern: `groups.setdefault(key, []).append(value)`.
- Interview choice: use a set when only presence matters; use a dictionary when you need counts or associated data.

```python
values = [2, 3, 2, 5, 3, 2]
unique_values = set(values)
counts = {}
for value in values:
    counts[value] = counts.get(value, 0) + 1
print(unique_values, counts)

student = ("Asha", 91)
name, score = student
print(name, score)
```

## 9. Reusable Interview Patterns

1. **Frequency map:** count values with a dictionary. Useful for duplicates, anagrams, and first unique items.
2. **Two pointers:** move left and right through a sorted list or string; often reduces nested loops to $O(n)$.
3. **Running maximum/minimum:** initialize from the first item, not always from `0` because all values may be negative.
4. **Prefix/suffix thinking:** precompute information before or after an index when repeated range work appears.
5. **Sort then scan:** sorting costs $O(n og n)$ but often makes duplicate and interval logic simple.
6. **Guard clauses:** handle empty input, one item, invalid ranges, and boundary values before the main logic.

### Complexity quick guide
- One pass over `n`: $O(n)$ time.
- Nested independent loops: often $O(n^2)$.
- Binary-style halving: $O(og n)$.
- Extra list/dictionary/set holding `n` items: $O(n)$ space.
- Always state whether sorting or copied slices add time or memory.

## 10. Common Bugs and Final Checklist

- `input()` is a string: cast it before comparing or adding numbers.
- `range(n)` stops before `n`; use `range(1, n + 1)` for 1 through n.
- Never use `is` for ordinary value comparison.
- Check empty lists before using index `0` or taking modulo of their length.
- Do not mutate a list while iterating over it unless you deliberately know the effect.
- `list.sort()` returns `None`; use `sorted_list = sorted(items)` when you need a result.
- Validate password conditions after scanning the whole password, not inside the loop on every character.
- For largest/second-largest problems, define how duplicates should behave.

### Interview checklist
1. Restate the problem and clarify assumptions.
2. Give a small example and identify edge cases.
3. Explain the approach before coding.
4. Use meaningful names and small helper functions.
5. Dry-run the code, then state time and space complexity.

```python
# A compact practice block: validate a password in one pass.
def valid_password(password):
    if len(password) < 8:
        return False
    return (
        any(character.isupper() for character in password)
        and any(character.islower() for character in password)
        and any(character.isdigit() for character in password)
        and any(not character.isalnum() for character in password)
    )

print(valid_password("Aiml@1234"))
```
